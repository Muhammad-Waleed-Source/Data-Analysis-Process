### Starting with Cleaning

### Before cleaning lets make a copy

```python
df1 = df.copy()
df1
```

### Working with price col

```python
# Removing the '₹' symbol

df1['price'] = df1['price'].str.replace('₹', '')
df1['price']
```

```python
# Now removing the commas
df1['price'] = df1['price'].str.replace(',', '')
df1['price']
```

```python
# as the price col is a string type so converting it to integer to free some memory

df1['price'] = df1['price'].astype('int')
df1.info()
```

### Resetting index

```python
# resetting the index so that we align with our observations

df1
```

```python
df1 = df1.reset_index()
df1
```

```python
df1['index'] = df1['index'] + 2
```

```python
df1

# now the original and copied data start from same index
```

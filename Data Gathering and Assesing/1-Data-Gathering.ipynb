### Opening a csv file from an URL

```python
import requests
from io import StringIO

url = "https://raw.githubusercontent.com/cs109/2014_data/master/countries.csv"
headers = {"User-Agent": "Mozilla/5.0 (Macintosh; Intel Mac OS X 10.14; rv:66.0) Gecko/20100101 Firefox/66.0"}
req = requests.get(url, headers=headers)
data = StringIO(req.text)

pd.read_csv(data)
```

### squeeze Method
When we want only one columns means it converts the dataframe into a series

```python
temp = pd.read_csv('Datasets/1-Data-Gathering/aug_train.csv', usecols=['gender']).squeeze()
print(temp)
print(type(temp)) # series
```

### skiprows/nrows Parameter

```python
pd.read_csv('Datasets/1-Data-Gathering/aug_train.csv').head()
```

```python
pd.read_csv('Datasets/1-Data-Gathering/aug_train.csv', skiprows=[0, 1]) # first two rows are skipped
```

```python
# we can also use our own logic inside the skiprows parameter using lambda
pd.read_csv('Datasets/1-Data-Gathering/aug_train.csv', skiprows=lambda x: x % 2 == 0 and x != 0) # for even rows
```

```python
# nrows (Number of rows needed)
pd.read_csv('Datasets/1-Data-Gathering/aug_train.csv', nrows=100)
```

### encoding Parameter

Some CSV files are saved using encodings other than UTF-8 (such as `latin-1`, `ISO-8859-1`, or `cp1252`). In such cases, we use the `encoding` parameter in `pd.read_csv()` to tell pandas how to correctly read the file.

```python
pd.read_csv('Datasets/1-Data-Gathering/zomato.csv', encoding='latin-1').head()
```

### on_bad_lines (Skip bad lines)
A bad line is a row that doesn't match the expected format. For example, if the CSV should have 3 columns, but one row has 4 columns, that row is considered bad.

```python
pd.read_csv('Datasets/1-Data-Gathering/BX-Books.csv', sep=';', encoding='latin-1', on_bad_lines='skip')
```

### Handling Dates (parse_dates)

```python
pd.read_csv('Datasets/1-Data-Gathering/IPL Matches 2008-2020.csv').head()

# here the dates are strings
```

```python
# now to convert them to date we will use parse_date parameter

pd.read_csv('Datasets/1-Data-Gathering/IPL Matches 2008-2020.csv', parse_dates=['date']).info()
```

### convertors Parameter
used in file-reading functions like pandas.read_csv that accepts a dictionary mapping column names or indices to custom function

```python
def rename(name):
    if name == 'Royal Challengers Bangalore':
        return 'RCB'
    else:
        return name
```

```python
rename('Royal Challengers Bangalore')
```

```python
pd.read_csv('Datasets/1-Data-Gathering/IPL Matches 2008-2020.csv', converters={'team1': rename}).head()
```

### na_values Parameter
to specify additional strings, numbers, or symbols that should be recognized as missing data (NaN)

```python
# example if we specify the Male as na value

pd.read_csv('Datasets/1-Data-Gathering/aug_train.csv', na_values=['Male']).head()
```

# Exporting Files

### 1. CSV

```python
df = pd.read_csv('Datasets/1-Data-Gathering/deliveries.csv')
df.head()
```

```python
temp_df = df.groupby('batsman')['batsman_runs'].sum().reset_index()

temp_df
```

```python
temp_df.to_csv('Datasets/1-Data-Gathering/batsman_runs.csv', index=False)
```

```python
temp_df2 = df.pivot_table(index='batsman', columns='bowling_team', values='batsman_runs', aggfunc='sum')
temp_df2
```

```python
temp_df2.to_csv('Datasets/1-Data-Gathering/batsman_vs_teams.csv')
```

### 2. Excel

```python
temp_df
```

```python
temp_df2
```

```python
temp_df.to_excel('Datasets/1-Data-Gathering/batsman_runs.xlsx')
```

```python
# exporting with a sheet name inside an excel file
temp_df.to_excel('Datasets/1-Data-Gathering/output2.xlsx', sheet_name='runs')
```

```python
# exporting in multiple sheets inside a single excel file
with pd.ExcelWriter('Datasets/1-Data-Gathering/output.xlsx') as writer:
    temp_df.to_excel(writer, sheet_name='Sheet_name_1')
    temp_df2.to_excel(writer, sheet_name='Sheet_name_2')
```
